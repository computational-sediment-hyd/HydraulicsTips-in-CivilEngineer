## 河床波と流水抵抗の話

### motivation

過去記事(Manningの粗度係数を再考する)に少し書きましたが、
河床波による流水抵抗に関する既往研究について改めて整理しました。

[Engelund](https://ascelibrary.org/doi/10.1061/JYCEAJ.0001417)によると、
流水に対する全せん断力$\tau_*$は、
摩擦抵抗によるせん断力$\tau_*^{\prime}$
と形状抵抗によるせん断$\tau_*^{\prime\prime}$の和で表すことができる(勾配分割法)。

$$
\begin{align}
 \tau_* = \tau_*^{\prime} + \tau_*^{\prime\prime}
\end{align}
$$

さらに、相似則より$\tau_*$と$\tau_*^{\prime}$の関係について下図(Engelund図と定義)を示した。
この図は、河床波形状によって$\tau_*$に対する$\tau_*^{\prime}$が変化することを示している。


```python
IFrame('https://computational-sediment-hyd.github.io/HydraulicsTips-in-CivilEngineer/fig/Engelund-taue.png', width=600, height=350)
```

その後の研究で$\tau_*$と$\tau_*^{\prime}$の関係は径深粒径比$R/d$によって変化することが知られている。

そこで、流水抵抗に関する研究である
マニング則、[岸・黒木の研究](https://www.jstage.jst.go.jp/article/jscej1969/1973/210/1973_210_65/_pdf/-char/ja)
、[芦田・道上の方法](https://www.jstage.jst.go.jp/article/jscej1969/1972/206/1972_206_59/_pdf)について
同図との比較を行った。

### マニング則とEngelund図の関係

Engelund図との比較にあたり、マニング則による$\tau_*^{\prime}$はEngelundと同様に以下の式を用いた。

$$
\begin{align}
\dfrac{V}{\sqrt{gR^{\prime}i_e}} &= \dfrac{1}{\kappa}\log_e{\dfrac{R^{\prime}}{k_s}} - \dfrac{1}{\kappa} + A_r   \\
k_s &=  2d_m \\
\tau_*^{\prime} &= \dfrac{R^{\prime}i_e}{\rho_{sw} d_m}
\end{align}
$$

マニングの粗度係数$n$は、0.01,0.02,0.03.0.04の4ケースで、
径深粒径比$R/d$と粒径$d$は以下の組み合わせの4ケースを計算した。
 - 100、10mm
 - 1000、10mm
 - 1000、1mm
 - 10000、1mm

結果は下図のとおりとなっている。

```python
IFrame('https://computational-sediment-hyd.github.io/HydraulicsTips-in-CivilEngineer/SandWavesAndResistanceToFlow/manning.html', width=650, height=480)
```

マニング則について考察する。

 - 河床波形状が変化する場合はマニングの粗度係数が一定では$\tau_*$と$\tau_*^{\prime}$の関係を表現できない。
 - dune領域では河床波が変化しなくてもマニングの粗度係数が一定とすると、$\tau_*$が1を超えるあたりからEngelund図との差が大きくなる。
 - flat-bed領域ではマニングの粗度係数を一定として計算可能である。

### 岸・黒木の研究とEngelund図の関係

岸・黒木の方法では、$\tau_*$と$\tau_*^{\prime}$の関係は$R/d$の関数として示される。
詳細は過去記事を参照
$R/d$は100,1000,10000の3ケースを図化した。

```python
IFrame('https://computational-sediment-hyd.github.io/HydraulicsTips-in-CivilEngineer/SandWavesAndResistanceToFlow/kishikuroki.html', width=650, height=380)
```

 - 岸・黒木の方法は、Engelund図が省略しているtransition領域をモデル化している。
 - $R/d$によって河床波形態が遷移する掃流力が異なることを示しているが、$R/d$が10000ではEngelund図と大きく乖離している。私の経験上もこの辺りは岸・黒木の方法の課題と感じる。


### 芦田・道上の方法とEngelund図の関係

掃流砂量式の芦田・道上式における有効掃流力を$\tau_*^{\prime}$として取り扱う方法を芦田・道上の方法と呼ぶ。

計算式は次のとおりである。

$$
\begin{align}
\dfrac{V}{u^{\prime}_{*}} &= \dfrac{1}{\kappa}\log_e{\dfrac{R}{k_s}} - \dfrac{1}{\kappa} + A_r   \\
k_s &=  d_m(1+2\tau_*) \\
\tau_*^{\prime} &= \dfrac{{u^{\prime}_*}^2}{\rho_{sw}g d_m}
\end{align}
$$

マニング則の場合と同様に、マニングの粗度係数$n$は、0.01,0.02,0.03.0.04の4ケースで、
径深粒径比$R/d$と粒径$d$は以下の組み合わせの4ケースを計算した。
 - 100、10mm
 - 1000、10mm
 - 1000、1mm
 - 10000、1mm

結果は下図のとおりとなっている。


```python
IFrame('https://computational-sediment-hyd.github.io/HydraulicsTips-in-CivilEngineer/SandWavesAndResistanceToFlow/AM.html', width=700, height=480)
```

マニング則とほぼ同様の関係を示している。
 - d=10mmではマニング則よりも$\tau_*$に対して$\tau_*^{\prime}$が大きくなっている。
 - d=1mmではマニング則と同程度になっている。
 
対数則を基本として、$\tau_*$の影響を考慮しているため、$\tau_*$が大きい（粒径が小さい）場合にマニング則との差が生じている。

### まとめ

 - 河床波形状が変化する場合は、マニング則、芦田・道上の方法では、マニングの粗度係数が一定では適切に抵抗を評価できない。
 - 岸・黒木の方法は河床波の変化を考慮できるが、$R/d=10000$ではEngelund図と大きく乖離する。

以上、古典的な研究のレビューをしてみました。近年の観測技術をふまえた新たな研究が待たれます。

##  create graphs

### module 

```python
import numpy as np
import pandas as pd
import holoviews as hv
from scipy.optimize import bisect 
from holoviews.plotting.util import process_cmap
cmap_list = process_cmap("glasbey_hv")

hv.extension('bokeh')
```

####  Engelund図

```python
df = pd.read_csv('https://computational-sediment-hyd.github.io/HydraulicsTips-in-CivilEngineer/data/Engelund.csv', header=None)
df.columns = ['x','y']

taus1 = np.logspace(-1,0.2,num=100,endpoint=True,base=10.0)
taus2 = np.logspace(-1.3,1.3,num=100,endpoint=True,base=10.0)

gbase = hv.Curve((df.x,df.y), label='Engelund').options(color='k',line_width=4) \
*hv.Curve((0.06+0.4*taus1**2,taus1), label='Engelund').options(color='k',line_width=4) \
*hv.Curve((0.4*taus1**2,taus1), label='τ*d=0.4τ*^2').options(color='k', line_dash='dotted',line_width=4) \
*hv.Curve((taus2,taus2), label='τ*d=τ*').options(color='k', line_dash='dashed',line_width=4) 
```

#### マニング

```python
def manningtaud(tauss, Rbyd=1000, d = 0.01, n=0.03):
    def subtaud(taus):
        g = 9.8
        Ar = 8.5
        ks = 2*d 
        kappa = 0.4
        ie = taus/Rbyd/1.65
        R = Rbyd*d
        U = ie**0.5*R**(2/3)/n
        
        def func(Rd):
            Rmin = ks*np.exp(1)
            if Rmin > Rd : 
                f = U/np.sqrt(g*Rd*ie) - Ar
            else :
                f = U/np.sqrt(g*Rd*ie) - (np.log(Rd/ks)-1)/kappa - Ar
                
            return f
        
        return bisect(func, 0.001, 100)*ie/1.65/d

    return np.array([subtaud(p) for p in tauss])
```

#### 岸・黒木 

```python
A1, A2 = 0.02, 0.07 
C1 = 0.21
C2 = C1/A1**0.5
C3 = A2**0.5
C4 = C1/A1**5

boundRbyd = (C4/(A2)**(-4.5))**(10/7)

def kishikuroki2(taus, Rbyd):
    
    if taus < A1*(Rbyd)**0.5:
        tausd = C1*taus**0.5
        tausd = tausd if tausd < taus else taus
    elif taus < A2*(Rbyd)**0.4:
        tausd = C2*(Rbyd)**(-0.25)*taus
        tausd = tausd if tausd < taus else taus
    else:
        tausd = C3*(Rbyd)**0.2*taus**0.5
        tausd = tausd if tausd < taus else taus
        
    return tausd

def kishikuroki1(taus, Rbyd):
    
    if taus < A1*(Rbyd)**0.5:
        tausd = C1*taus**0.5
        tausd = tausd if tausd < taus else taus
    elif taus < A2*(Rbyd)**0.4:
        tausd = C4*(Rbyd)**(-2.5)*taus**5.5 
        tausd = tausd if tausd < taus else taus
    else:
        tausd = C3*(Rbyd)**0.2*taus**0.5
        
    return tausd

def figKishiKuroki(taus, Rbyd, color):
    tausd1 = [ kishikuroki1(tt, Rbyd) for tt in taus ]
    tausd2 = [ kishikuroki2(tt, Rbyd) for tt in taus ]
    g1 = hv.Curve((tausd1, taus2),label='R/d='+str(Rbyd)).options(color=color, line_width=2, show_grid=True)
    g2 = hv.Curve((tausd2, taus2),label='R/d='+str(Rbyd)).options(color=color, line_width=2, show_grid=True)
    return g1*g2
```

#### 芦田・道上 

```python
def amtaud(tauss, Rbyd, d, n):
    
    def subtaud(taus):
        g = 9.8
        Ar = 8.5
        kappa = 0.4
    #     ks = 2*d 
        ie = taus/Rbyd/1.65
        R = Rbyd*d
        U = ie**0.5*R**(2/3)/n
        
        if np.log(Rbyd/(1.0 + 2.0*taus)) < 1.0:
            phi = Ar
        else:
            phi = (np.log(Rbyd/(1+2.0*taus))-1)/kappa + Ar
        
        Ue = U/phi
        taue = Ue**2/1.65/g/d
        
        return taue

    return np.array([subtaud(p) for p in tauss])
```

###  マニングの式


```python
def mkfig(Rbyd, d):
    g = gbase \
    *hv.Curve((manningtaud(taus2, Rbyd, d, n=0.01), taus2),label='n=0.010').options(color = cmap_list[0]) \
    *hv.Curve((manningtaud(taus2, Rbyd, d, n=0.02), taus2), label='n=0.020').options(color = cmap_list[1]) \
    *hv.Curve((manningtaud(taus2, Rbyd, d, n=0.03), taus2), label='n=0.030').options(color = cmap_list[2]) \
    *hv.Curve((manningtaud(taus2, Rbyd, d, n=0.04), taus2), label='n=0.040').options(color = cmap_list[3]) 
    
    gl = g.options(height=350, width=600, logx=True, logy=True, legend_position='right'
              , xlabel='τ*d', ylabel='τ*',show_grid=True).redim.range(x=(0.01,10), y=(0.05,10))
    return gl
```

```python
dic = {}
dic['R/d=100, d=10mm'] = mkfig(100, 0.01)
dic['R/d=1000, d=10mm'] = mkfig(1000, 0.01)
dic['R/d=1000, d=1mm'] = mkfig(1000, 0.001)
dic['R/d=10000, d=1mm'] = mkfig(10000, 0.001)

hmap = hv.HoloMap(dic, kdims=['param'])
```
